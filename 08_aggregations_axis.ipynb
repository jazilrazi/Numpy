```python
#1. The problem: sum of the whole grid

import numpy as np
grid = np.array([[1,2,3],
                 [4,5,6]])
print(np.sum(grid))
```

```python
#2. axis=0 — sum DOWN each column

print(np.sum(grid,axis=0))


#Column 1: 1+4=5, Column 2: 2+5=7, Column 3: 3+6=9.
```

```python
#3. axis=1 — sum ACROSS each row

print(np.sum(grid,axis=1))


#Row 1: 1+2+3=6, Row 2: 4+5+6=15.
```

```python
#4. Works with other functions too
print (np.mean(grid,axis=0))
print(np.max(grid,axis=1))
```
