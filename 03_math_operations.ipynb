```python
#1. Math directly on an array

import numpy as np
arr = np.array([10,20,30,40])

print(arr + 5)
print(arr - 5)
print(arr * 5)
print(arr / 5)
```

```python
#2. Math between two arrays (same size)

a = np.array([1,2,3,])
b = np.array([10,20,30])

print(a * b)
print(a + b)
print(a - b)
print(a / b)
```

```python
#3. Useful math functions

nums = np.array([10,20,30,40,50])

print(np.sum(nums))
print(np.mean(nums))
print(np.min(nums))
print(np.max(nums))
```
