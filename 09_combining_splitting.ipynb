```python
#1. np.concatenate() — join arrays end to end

import numpy as np
a = np.array([1,2,3])
b = np.array([4,5,6])
result = np.concatenate((a,b))
print (result)

#Just glues them together, one after the other.
```

```python
import numpy as np
a=np.array([1,2,3])
b=np.array([4,5,6])
vertical_stack=result = np.vstack((b,a))
#"v" = vertical. Think of it like stacking two rows on top of each other to make a grid.
horizondel_stack=result = np.hstack((b,a))
#"h" = horizontal. For 1D arrays this looks the same as concatenate, but it becomes clearer with 2D arrays:
print(horizondel_stack)
print(vertical_stack)

```
