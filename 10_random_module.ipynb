```python
#1. Random decimal numbers between 0 and 1

import numpy as np
r = np.random.rand(3)
print(r)
# rand(3) gives you 3 random decimal numbers, each between 0 and 1. The exact numbers will be different every time you run it.
```

```python
#2. Random whole numbers (integers) in a range

num = np.random.randint(0,10,size=5)
print(num)

# This means: "give me 5 random whole numbers between 1 and 9" (10 is not included, just like slicing).
```

```python
# 3. Random numbers arranged in a grid
r = np.random.rand(2,3)
print(r)

# Same idea, but shaped into 2 rows and 3 columns.
```

```python
# 4. np.random.seed() — make randomness repeatable

r = np.random.seed(40)
print(np.random.randint(0,10,size=3))

# Normally random numbers are different every time. But if you set a seed,
#  you'll get the same "random" numbers every time — very useful so your code gives consistent,
#  testable results.
```
