```python
#First, we always need this line to use NumPy
import numpy as np
#This just means "bring in NumPy and let me call it np for short."
```

```python
#1. Make an array from a list
arr = np.array([1,2,3,4])
print(arr)
```

```python
#2. Make an array of all zeros
zeros = np.zeros(5)
print(zeros)
```

```python
#3. Make an array of all ones
ones = np.ones(4)
print(ones)
```

```python
#4. Make an array with a range of numbers
r = np.arange(1,6)
print(r)
```

```python
#5. Make evenly spaced numbers
lin = np.linspace(0, 10, 5)
print(lin)
```
